```python
import pandas as pd
import string
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
```

```python
data = {
    "Review ID": [1,2,3,4,5,6,7,8],
    "Customer Review": [
        "The laptop is fast and the battery lasts long.",
        "The battery life is excellent and performance is impressive.",
        "The laptop is slow but the display quality is good.",
        "Excellent performance and excellent battery backup.",
        "The display is bright and the laptop looks premium.",
        "Battery performance is poor and charging is slow.",
        "The laptop has excellent speed and bright display.",
        "Poor battery life but excellent display quality."
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Tokenize
    words = word_tokenize(text)

    # Remove stopwords
    words = [word for word in words if word not in stop_words]

    return " ".join(words)

df["Processed Review"] = df["Customer Review"].apply(preprocess)

print(df[["Customer Review","Processed Review"]])
```

```python
count_vectorizer = CountVectorizer()

count_matrix = count_vectorizer.fit_transform(df["Processed Review"])

count_df = pd.DataFrame(
    count_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

print("Vocabulary")
print(count_vectorizer.vocabulary_)

print("\nFeature Names")
print(count_vectorizer.get_feature_names_out())

print("\nDocument-Term Matrix")
print(count_df)
```

```python
binary_vectorizer = CountVectorizer(binary=True)

binary_matrix = binary_vectorizer.fit_transform(df["Processed Review"])

binary_df = pd.DataFrame(
    binary_matrix.toarray(),
    columns=binary_vectorizer.get_feature_names_out()
)

print(binary_df)
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(df["Processed Review"])

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("Vocabulary")
print(tfidf_vectorizer.vocabulary_)

print("\nFeature Names")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix")
print(tfidf_df.round(3))
```
