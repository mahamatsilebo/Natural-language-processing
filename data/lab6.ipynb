```python
# Install required libraries (Run once)
# !pip install nltk gensim pandas

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from gensim.models import Word2Vec
import string

# Download required NLTK resources (Run once)
nltk.download('punkt')
nltk.download('stopwords')
```

```python


# -------------------------------------
# 1. Read, preprocess and tokenize data
# -------------------------------------

with open("/Users/mahamatsilebo/Downloads/campus_corpus.txt", "r", encoding="utf-8") as file:
    text = file.read().lower()

# Remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

# Tokenize sentences
sentences = text.split("\n")

# Stop words
stop_words = set(stopwords.words("english"))

processed_sentences = []

for sentence in sentences:
    words = word_tokenize(sentence)
    words = [word for word in words if word not in stop_words]
    processed_sentences.append(words)

print("Tokenized Corpus:")
for sentence in processed_sentences:
    print(sentence)
```

```python

# -------------------------------------
# 2. Train Word2Vec Model
# -------------------------------------

model = Word2Vec(
    sentences=processed_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=1      # Skip-Gram
)
```

```python

# -------------------------------------
# 3. Print Vocabulary
# -------------------------------------

print("\nVocabulary Learned:")
print(list(model.wv.index_to_key))
```

```python

# -------------------------------------
# 4. Display Vector Representation
# -------------------------------------

word = "machine"

print(f"\nVector representation of '{word}':")

if word in model.wv:
    print(model.wv[word])
else:
    print("Word not found.")
```

```python

# -------------------------------------
# 5. Find Most Similar Words to "machine"
# -------------------------------------

print("\nWords most similar to 'machine':")

if "machine" in model.wv:
    similar_words = model.wv.most_similar("machine", topn=5)

    for w, score in similar_words:
        print(f"{w} : {score:.4f}")
else:
    print("Word not found.")
```

```python

# -------------------------------------
# 6. Similarity between
#    artificial and intelligence
# -------------------------------------

word1 = "artificial"
word2 = "intelligence"

print(f"\nSimilarity between '{word1}' and '{word2}':")

if word1 in model.wv and word2 in model.wv:
    similarity = model.wv.similarity(word1, word2)
    print(similarity)
else:
    print("One or both words not found.")
```

```python

# -------------------------------------
# 7. Find Odd Word
# -------------------------------------

words = ["python", "machine", "banana", "algorithm"]

print("\nOdd word:")
print(model.wv.doesnt_match(words))
```
