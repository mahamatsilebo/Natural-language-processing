```python
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")

df = pd.read_csv("/Users/mahamatsilebo/Downloads/structured_agriculture_data.csv")

df["POS"] = df["Word"].apply(lambda x: nlp(str(x))[0].pos_)

print(df.head())
```

```python
print(df["POS"].value_counts())
```

```python
nouns = df[df["POS"]=="NOUN"]

print(nouns.sort_values("Frequency", ascending=False).head(20))
```

```python
verbs = df[df["POS"]=="VERB"]

print(verbs.sort_values("Frequency", ascending=False).head(20))
```

```python
adjs = df[df["POS"]=="ADJ"]

print(adjs.sort_values("Frequency", ascending=False).head(20))
```

```python
entities = []

for word in df["Word"]:
    doc = nlp(str(word))
    for ent in doc.ents:
        entities.append((ent.text, ent.label_))

print(entities)
```
